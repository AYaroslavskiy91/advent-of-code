## Part A

```python
inp = examples[0].input_data
inp
```

```python
lines = inp.split("\n")
```

```python
def flip_switches(all_switches, length):
    result = defaultdict(bool)
    for l in range(length):
        result[l]
    for nums in all_switches:
        switches = [int(d) for d in re.findall(r"\d", nums)]
        for d in switches:
            result[d] = not result[d]
    return result
```

```python
def get_final(final_raw):
    final = defaultdict(bool)
    for idx, item in enumerate(final_raw.strip("[]")):
        if item=="#":
            final[idx]=True
        else:
            final[idx]=False
    return final
```

```python
def try_combos(switches, final):
    result = get_final(final)
    length = len(result)
    for max_switches in range(1, len(switches)):
        for combo in combinations(switches, max_switches):
            attempt = flip_switches(combo, length)
            if result==attempt:
                return max_switches
```

```python
presses = 0
for line in lines:
    parts = line.split(" ")
    final = parts[0]
    joltage = parts[-1]
    switches = parts[1:-1]

    try:
        presses += try_combos(switches, final)
    except:
        print(line)
presses
```

```python
answer_a = presses
```

## Part A Submission

```python
puzzle.answer_a = answer_a
```

## Part B

```python
def flip_switches_b(all_switches, length):
    result = defaultdict(int)
    for l in range(length):
        result[l]
    for nums in all_switches:
        switches = [int(d) for d in re.findall(r"\d", nums)]
        for d in switches:
            result[d] += 1
    return result
```

```python
def get_final_b(joltage):
    final = defaultdict(int)
    for idx, item in enumerate(joltage.strip("{}").split(",")):
        final[idx] = int(item)
    return final
```

```python
def try_combos_b(switches, final):
    max_presses = sum([int(x) for x in re.findall(r"\d+", final)])
    result = get_final_b(final)
    length = len(result)
    print(max_presses, result)
    for max_switches in range(1, len(switches)):
        for combo in combinations(switches, max_switches):
            for max_switches in range(len(combo), max_presses):
                for multi_combo in combinations_with_replacement(combo, max_presses):
                    attempt = flip_switches_b(multi_combo, length)
                    print(attempt)
                    if result==attempt:
                        return max_switches
```

```python
presses_b = 0
for line in lines:
    parts = line.split(" ")
    final = parts[0]
    joltage = parts[-1]
    switches = parts[1:-1]

    try:
        presses += try_combos_b(switches, joltage)
    except:
        print(line)
```

```python
answer_b = presses_b
```

## Part B Submission

```python
puzzle.answer_b = answer_b
```
