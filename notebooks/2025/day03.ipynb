```python
from aocd.models import Puzzle
import numpy as np
import re
```

## Get Puzzle Input

```python
year = 2025
day = 3
puzzle = Puzzle(year=year, day=day)
```

```python
inp = puzzle.input_data
```

```python
examples = puzzle.examples
examples
```

```python
url = puzzle.url
url
```

## Part A

```python
def get_max_jolts(batteries):
    for num in np.arange(9, 0, -1):
        if str(num) in batteries[:-1]:
            idx = re.search(str(num), batteries).span()[0]
            second = np.max([int(x) for x in re.findall(r'(\d)', batteries[idx+1:])])
            return 10*num+second
```

```python
jolts = 0
for bats in inp.split('\n'):
    jolts += get_max_jolts(bats)
jolts
```

```python
answer_a = jolts
```

## Part A Submission

```python
puzzle.answer_a = answer_a
```

## Part B

```python
def get_max_jolts_b(batteries, length):
    if length>1:
        for num in np.arange(9, 0, -1):
            if str(num) in batteries[:-length+1]:
                idx = re.search(str(num), batteries).span()[0]
                second = get_max_jolts_b(batteries[idx+1:], length-1)
                return batteries[idx]+second
    for num in np.arange(9, 0, -1):
        if str(num) in batteries:
            idx = re.search(str(num), batteries).span()[0]
            return batteries[idx]
```

```python
jolts = 0
for bats in inp.split('\n'):
    jolts += int(get_max_jolts_b(bats, 12))
jolts
```

```python
answer_b = jolts
```

## Part B Submission

```python
puzzle.answer_b = answer_b
```
