## Part A

```python
def pad(grid):
    length = len(grid[0])+2
    padded = ['.'*length]
    
    for line in grid:
        padded.append('.'+line+'.')
    padded.append('.'*length)

    return padded
```

```python
def get_neighbors(idx_y, idx_x):
    return [
        (idx_y+1, idx_x+1),
        (idx_y+1, idx_x),
        (idx_y+1, idx_x-1),
        (idx_y-1, idx_x+1),
        (idx_y-1, idx_x),
        (idx_y-1, idx_x-1),
        (idx_y, idx_x+1),
        (idx_y, idx_x-1),
    ]
```

```python
def is_accessible(idx_y, idx_x):
    neighbors = get_neighbors(idx_y, idx_x)
    cnt = 0
    for (y, x) in neighbors:
        if padded[y][x]=="@":
            cnt+=1
    if cnt<4:
        return 1
    return 0
```

```python
def get_roll_count(padded):
    accessible = 0

    for idx_y, line in enumerate(padded):
        if idx_y==0 or idx_y == len(padded):
            continue
        for idx_x, item in enumerate(line):
            if idx_x==0 or idx_x == len(padded) or item==".":
                continue
    
            accessible+=is_accessible(idx_y, idx_x)
    return accessible
```

```python
grid = inp.split('\n')
padded = pad(grid)
answer_a = get_roll_count(padded)
answer_a
```

## Part A Submission

```python
puzzle.answer_a = answer_a
```

## Part B

```python
def get_roll_count_and_grid(padded):
    accessible = 0
    idc = []
    for idx_y, line in enumerate(padded):
        if idx_y==0 or idx_y == len(padded):
            continue
        for idx_x, item in enumerate(line):
            if idx_x==0 or idx_x == len(padded) or item==".":
                continue

            if is_accessible(idx_y, idx_x):
                accessible+=1
                idc.append((idx_y, idx_x))

    new_pad = padded.copy()
    for idx in idc:
        # print(idx)
        new_pad[idx[0]]=new_pad[idx[0]][:idx[1]] + '.' + new_pad[idx[0]][idx[1]+1:]
    return new_pad, accessible
```

```python
grid = inp.split('\n')
padded = pad(grid)
full_accessible = 0

go_time = True
while go_time:
    padded, accessible = get_roll_count_and_grid(padded)
    full_accessible += accessible
    if accessible==0:
        go_time=False
    
```

```python
answer_b = full_accessible
```

## Part B Submission

```python
puzzle.answer_b = answer_b
```
