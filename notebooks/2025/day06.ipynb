```python
from aocd.models import Puzzle
import pandas as pd
import numpy as np
import re
from itertools import groupby
```

## Get Puzzle Input

```python
year = 2025
day = 6
puzzle = Puzzle(year=year, day=day)
```

```python
inp = puzzle.input_data
```

```python
examples = puzzle.examples
examples
```

```python
url = puzzle.url
url
```

## Part A

```python
lines = inp.split('\n')
```

```python
addends = pd.DataFrame([re.findall(r'\d+', item) for item in lines[:-1]]).T.astype(int).assign(
    operations = re.findall(r'\*|\+', lines[-1])
)
sum_mask = addends.operations.eq("+")
total = addends.assign(
    computed = lambda x: np.where(sum_mask, np.sum(addends.drop(columns="operations"), axis=1), np.prod(addends.drop(columns="operations"), axis=1))
).computed.sum()
```

```python
answer_a = total
```

## Part A Submission

```python
puzzle.answer_a = answer_a
```

## Part B

```python
subs = []
ops = re.findall(r'\*|\+', lines[-1])
ops.reverse()

transposed = ["".join(row) for row in np.array([[char for char in line] for line in lines[:-1]]).T]
new_lines = [
    [item.strip() for item in group]
    for mask, group in groupby(transposed, lambda x: bool(x.strip()))
    if mask
]
new_lines

total = 0
for line, op in zip(new_lines, ops[::-1]):
    if op=="+":
        sol = np.sum([int(item) for item in line])
    else:
        sol = np.prod([int(item) for item in line])
    total += sol
total
```

```python
answer_b = total
```

## Part B Submission

```python
puzzle.answer_b = answer_b
```
