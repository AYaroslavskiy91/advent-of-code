```python
from aocd.models import Puzzle
import numpy as np
import pandas as pd
```

## Get Puzzle Input

```python
year = 2025
day = 1
puzzle = Puzzle(year=year, day=day)
```

```python
inp = puzzle.input_data
```

```python
examples = puzzle.examples
examples
```

## Part A

```python
def rotate_dial(inp_line):
    direction, magnitude = inp_line[0], inp_line[1:]
    match direction:
        case "R":
            return int(magnitude)
        case "L":
            return -int(magnitude)
```

```python
rots = inp.split('\n')
zeros = 0
dial = 50

for rot in rots:
    dial = (dial + rotate_dial(rot))%100
    if dial==0:
        zeros += 1
```

```python
answer_a = zeros
```

## Part A Submission

```python
puzzle.answer_a = answer_a
```

## Part B

```python
def get_num_zeros(delta, prev):
    new = prev+delta
    match (new, prev):
        case (new, prev) if new > 0:
            return int((delta+prev)/100)
        case (new, prev) if new < 0 and prev != 0:
            return int(np.abs(delta+prev)/100) + 1
        case (new, prev) if new < 0 and prev == 0:
            return int(np.abs(delta+prev)/100)
        case (new, prev) if new == 0:
            return 1
```

```python
rots = inp.split('\n')
zeros = 0
dial = 50

for rot in rots:
    prev = dial
    dial = (dial + rotate_dial(rot))%100
    
    zeros += get_num_zeros(rotate_dial(rot), prev)

    # print(dial, rotate_dial(rot), zeros)
```
