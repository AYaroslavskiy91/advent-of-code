```python
from aocd.models import Puzzle
import re
import numpy as np
from collections import defaultdict
```

## Get Puzzle Input

```python
year = 2025
day = 7
puzzle = Puzzle(year=year, day=day)
```

```python
inp = puzzle.input_data
```

```python
examples = puzzle.examples
examples
```

```python
url = puzzle.url
url
```

## Part A

```python
grid = inp.split("\n")
```

```python
start = re.search("S", grid[0]).span()[0]
start
```

```python
idxs = [start]
splits = 0
for line in grid[1:]:
    splitters = [x.span()[0] for x in re.finditer(r"\^", line)]
    to_remove = []
    to_add = []
    for idx in idxs:
        if idx in splitters:
            to_remove = [*to_remove] + [idx]
            to_add = [*to_add] + [idx+1, idx-1]
            splits += 1
    idxs = list(set(idxs) - set(to_remove) | set(to_add))
    # print(idxs, splitters)

splits
```

```python
answer_a = splits
```

## Part A Submission

```python
puzzle.answer_a = answer_a
```

## Part B

```python
idxs = defaultdict(int)
idxs[start] = 1
splits = 0
for line in grid[1:]:
    splitters = [x.span()[0] for x in re.finditer(r"\^", line)]
    deltas = defaultdict(int)
    for idx, cnt in idxs.items():
        if idx in splitters:
            deltas[idx]-=cnt
            deltas[idx+1]+=cnt
            deltas[idx-1]+=cnt
            splits += 1
    for idx, cnt in deltas.items():
        idxs[idx] += cnt
    # print(idxs, splitters)

num_timelines = np.sum([x for x in idxs.values()])
```

```python
answer_b = num_timelines
```

## Part B Submission

```python
puzzle.answer_b = answer_b
```
