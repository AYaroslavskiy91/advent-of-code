## Part A

```python
box_coords = [np.array([int(item) for item in line.split(",")]) for line in inp.split("\n") ]
df = (
    pd.DataFrame(combinations(box_coords, 2)).rename(columns={0:"box1", 1:"box2"})
    .assign(
        connected = 0
    )
)
df['dist'] = np.linalg.norm(np.vstack(df.box1)-np.vstack(df.box2), axis=1)
df.sort_values(by="dist", inplace=True)
df.reset_index(drop=True, inplace=True)
```

```python
def get_prod(df, cnt):
    df.loc[:cnt-1, "connected"]=1
    circuits = []
    
    for item in df[df.connected==1].itertuples():
        connection = {tuple(item.box1), tuple(item.box2)}
        circuits.append(connection)
        for circ in circuits:
            if (tuple(item.box1) in circ or tuple(item.box2) in circ) and circ!=connection:
                circuits.remove(circ)
                circuits.remove(connection)
                new_circuit = circ | connection
                circuits.append(new_circuit)
    
                for circ_2 in circuits:
                    if (not set.isdisjoint(new_circuit, circ_2)) and circ_2!=new_circuit:
                        circuits.remove(circ_2)
                        circuits.remove(new_circuit)
                        circuits.append(new_circuit | circ_2)
                        break
                break
    prod = np.prod(np.sort([len(x) for x in circuits])[-3:])
    num_circuits = len(circuits)
    return prod, num_circuits
```

```python
answer_a, num_circuits = get_prod(df, 1000)
```

## Part A Submission

```python
puzzle.answer_a = answer_a
```

## Part B

```python
def find_last_box(df):
    circuits = []
    
    for item in df.itertuples():
        connection = {tuple(item.box1), tuple(item.box2)}
        circuits.append(connection)
        for circ in circuits:
            if (tuple(item.box1) in circ or tuple(item.box2) in circ) and circ!=connection:
                circuits.remove(circ)
                circuits.remove(connection)
                new_circuit = circ | connection
                circuits.append(new_circuit)
    
                for circ_2 in circuits:
                    if (not set.isdisjoint(new_circuit, circ_2)) and circ_2!=new_circuit:
                        circuits.remove(circ_2)
                        circuits.remove(new_circuit)
                        circuits.append(new_circuit | circ_2)
                        break
                break
        if len(circuits)==1 and len(circuits[0])==1000:
            return circuits, item
    return circuits, item
```
