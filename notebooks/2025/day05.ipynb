```python
from aocd.models import Puzzle
import numpy as np
import re
```

## Get Puzzle Input

```python
year = 2025
day = 5
puzzle = Puzzle(year=year, day=day)
```

```python
inp = puzzle.input_data
```

## Part A

```python
def discretize_ranges_ingredients(ranges_split, ingredients):
    length = len(min(min(ranges_split)))
    for ran in ranges_split:
        m, M = ran[0], ran[1]
        cur_mlen = len(m)
        cur_Mlen = len(M)

        mdiff = length - cur_mlen
        Mdiff = length - cur_Mlen

        if mdiff>0:
            m = "0"*mdiff + m
        if Mdiff>0:
            M = "0"*Mdiff + M
        ranges_split[ranges_split.index(ran)] = [m, M]

    for ing in ingredients:
        cur_ilen = len(ing)

        idiff = length - cur_ilen
        if cur_ilen>0:
            update_ing = "0"*idiff + ing
        ingredients[ingredients.index(ing)]=update_ing
        
    ranges_split.sort()
    new_ranges = [ranges_split[0]]
    for ran in ranges_split[1:]:
        min_cur, max_cur = ran[0], ran[1]
        min_prev, max_prev = new_ranges[-1][0], new_ranges[-1][1]

        if min_cur <= max_prev and max_cur <= max_prev:
            continue;

        elif min_cur <= max_prev and max_cur > max_prev:
            new_ranges[-1][1] = max_cur

        else:
            new_ranges.append([min_cur, max_cur])

    

    return new_ranges, ingredients
```

```python
ranges, ingredients = [x.split('\n') for x in inp.split('\n\n')]
ranges_split = [re.findall(r'\d+', ran) for ran in ranges]
disc_ranges, disc_ingredients = discretize_ranges_ingredients(ranges_split, ingredients)
fresh = 0
for ing in disc_ingredients:
    ing_range = [ing, ing]
    ranges_ing = disc_ranges.copy() + [ing_range]
    ranges_ing.sort()
    index = ranges_ing.index(ing_range)
    if ranges_ing[index-1][0] <= ing and ranges_ing[index-1][1] >= ing:
        fresh += 1
        # print(ing)
fresh
```

```python
answer_a = fresh
```

## Part A Submission

```python
puzzle.answer_a = answer_a
```

## Part B

```python
total_possible_fresh = 0
for ran in disc_ranges:
    diff = int(ran[1]) - int(ran[0]) + 1
    total_possible_fresh += diff
total_possible_fresh
```

```python
answer_b = total_possible_fresh
```

## Part B Submission

```python
puzzle.answer_b = answer_b
```
