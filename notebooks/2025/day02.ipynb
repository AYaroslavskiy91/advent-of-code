## Part A

```python
def get_range(id_range):
    id_min, id_max = [int(x) for x in re.findall(r'\d+', id_range)]
    return id_min, id_max
```

```python
def check_invalid(num):
    num_str = str(num)
    return num_str[:int(len(num_str)/2)]==num_str[int(len(num_str)/2):]
```

```python
def solve_a(inp):
    invalids = 0
    for inp_range in inp.split(","):
        id_min, id_max = get_range(inp_range)
        for num in range(id_min, id_max+1):
            if check_invalid(num):
                invalids += num
    return invalids
```

```python
answer_a = solve_a(inp)
```

## Part A Submission

```python
puzzle.answer_a = answer_a
```
